# Case Study: A Machine Recognizing Handwritten Digits

A machine recognizing handwritten digits may not seem impressive since humans can perform this task really easy.

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/what_is_a_neural_network_01.png", width=500))
display(Markdown("*A human can tell instantly that these are all 3*"))
```

If we are tasked to write a program that takes image of handwritten digit (28x28 pixels) and outputs the digit 0 to 9, it becomes really difficult. Somehow, **tasks that are really easy for human brain is very hard to describe how to solve**. Hence, traditional computer programming based on if conditions, for loops, classes, objects and functions is not suited for this kind of problem.

Human brain has network of neurons and it learns from data, by observing numerous examples. **What if we create a brain-inspired software and hope that using this software, we can solve tasks like handwritten digit recognition which are easy to solve for human brain, but very difficult to describe how to solve?** This is called a **Neural Network**.

![rainbow](https://github.com/ancilcleetus/My-Learning-Journey/assets/25684256/839c3524-2a1d-4779-85a0-83c562e1e5e5)

# The Structure of a Neural Network

There are multiple variants of neural nets like Convolutional Neural Nets (CNNs), Recurrent Neural Nets (RNNs), Transformers etc. The first step to understand any of them is to build up the most fundamental form called Plain Vanilla aka Multi-Layer Perceptron (MLP).

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/what_is_a_neural_network_02.png", width=500))
display(Markdown("*The simple network we’re using to identify digits is just a few layers of neurons linked together.*"))
```

## Neurons

Neural networks are just a bunch of neurons connected together. You can visualize a neuron as "a thing that holds a number" (between 0.0 and 1.0 for now). This number is called "activation" of that neuron.

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/what_is_a_neural_network_03.png", width=500))
display(Markdown("*Every neuron has an activation between 0.0 and 1.0, sort of analogous to how neurons in the brain can be active or inactive.*"))
```

All of the information passed through the neural network is stored in these neurons. Hence, we need to represent the inputs (28x28 pixels) and outputs (digit prediction) in terms of neurons with values between 0.0 and 1.0.

To represent digital image with 28x28 pixel values, we create an input layer of 784 neurons.

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/what_is_a_neural_network_04.png", width=500))
display(Markdown("*Each pixel in the original image has a value between 0.0 (black) and 1.0 (white).*"))
```

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/what_is_a_neural_network_05.png", width=500))
display(Markdown("*The input layer contains 784 neurons, each of which corresponds to a single pixel in the original image.*"))
```

The last output layer has 10 neurons corresponding to digits 0-9. Activations of these 10 neurons describe the possibility of each digit w.r.t the neural network's prediction.

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/what_is_a_neural_network_06.png", width=500))
display(Markdown("*The output layer of our network has 10 neurons. Each neuron corresponds to a particular digit that the image could contain.*"))
```

## The Hidden Layers

There are some "hidden" layers between input & output layers. Here we have 2 hidden layers of 16 neurons each.

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/what_is_a_neural_network_07.png", width=500))
display(Markdown("*The Hidden Layers*"))
```
