# Case Study: A Machine Recognizing Handwritten Digits

A machine recognizing handwritten digits may not seem impressive since humans can perform this task really easy.

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/what_is_a_neural_network_01.png", width=500))
display(Markdown("*A human can tell instantly that these are all 3*"))
```

If we are tasked to write a program that takes image of handwritten digit (28x28 pixels) and outputs the digit 0 to 9, it becomes really difficult. Somehow, **tasks that are really easy for human brain is very hard to describe how to solve**. Hence, traditional computer programming based on if conditions, for loops, classes, objects and functions is not suited for this kind of problem.

Human brain has network of neurons and it learns from data, by observing numerous examples. **What if we create a brain-inspired software and hope that using this software, we can solve tasks like handwritten digit recognition which are easy to solve for human brain, but very difficult to describe how to solve?** This is called a **Neural Network**.
