# Introduction

- In a perfect world, a data scientist would possess "God's View" access to every data point, past, present and future. If you wanted to know the average height of humans, you would measure all 8 billion humans. If you wanted to know whether a user clicks an ad, you would simulate every possible user in existence.
- In reality, we are constrained by time, cost and physics. We cannot measure everything. We need to rely on a subset of reality to understand the whole. This is the core conflict of statistic - **The Truth (Population) vs The Evidence (Sample)**.
- **The Intuition** - Think of soup testing
    - **Population**: The entire pot of soup. To know the flavour of the soup, you would need to drink the entire pot (destroys the product).
    - **Sample**: A single spoonful
    - **Inference**: If the spoonful is salty, entire pot is salty

    *What if we didn't stir the pot? Even though spoonful is not salty, the bottom is pure salt. This problem is called **sampling bias**.*
