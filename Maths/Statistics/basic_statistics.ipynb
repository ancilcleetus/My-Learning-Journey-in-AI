# Intro to Descriptive Statistics

What is Descriptive Statistics?
- Descriptive Statistics => tools to summarize and describe data
- Instead of looking at thousands of numbers, we extract key metrics that tell the story
- Eg: "This app has a 4.5 rating based on 10,000 reviews" => Opinions of thousands of people summarized into a single useful metric.

Essential Building Blocks
- Where is the "center" of your data? => **mean**, **median**, **mode**
- How spread out is your data? => **variance**, **standard deviation**
- Symmetry of your data => **Skewness**
- Tails in your data => **Kurtosis**

Why this matters for ML Engineers?
- **Make sense of data**: Quickly summarize thousands of numbers into a few key insights.
- **Spot problems**: Identify unusual values which can be errors or exceptions.
- **Compare fairly**: Understand if two datasets are truly different or just appear so.
- **Build intuition**: These concepts are the foundations for everything in data science.

# The Mean (Average)

- The most common way to summarize data is to find its "center" - a single value representing a "typical" data point. Mean is the most used measure of center.
- **Arithmetic Mean** (aka Average) is the sum of all values divided by the number of values:

$$
\overline{x} = \frac {1}{n}\sum_{i=1}^{n} x_i = \frac {x_1 + x_2 + ... + x_n}{n}
$$

- **Example: Bulb Lifespans**

    You test 5 bulbs with lifespans: 980, 1000, 1010, 1020, 1040 hours.

    Mean = (980 + 1000 + 1010 + 1020 + 1040) / 5 = 1010 hours

![rainbow](https://github.com/ancilcleetus/My-Learning-Journey/assets/25684256/839c3524-2a1d-4779-85a0-83c562e1e5e5)

# The Median (Middle Value)

- The mean works well for symmetric data, but it has a weakness - outliers can pull it away from the true center. Median solves this weakness by **focusing on the position instead of the value**.
- Median is the middle value when data points are sorted. If we have even number of data points, median is the average of the two middle values:

$$
\text{Median} =
\begin{cases}
  x_{\frac{n+1}{2}}, & \text{if } n \text{ is odd} \\
  \frac{x_{\frac{n}{2}} + x_{\frac{n}{2} + 1}}{2}, & \text{if } n \text{ is even}
\end{cases}
$$

- **Example: With an Outlier**
    
    Lifespans: 500, 980, 1000, 1010, 1020 hours. (One bulb failed early.)

    Sorted: 500, 980, 1000, 1010, 1020
    
    Median: 1000 hours (middle value)
    
    Mean: 902 hours (pulled down by the outlier)

    *Notice how the median is more representative of the "typical" bulb.*

![rainbow](https://github.com/ancilcleetus/My-Learning-Journey/assets/25684256/839c3524-2a1d-4779-85a0-83c562e1e5e5)

# The Mode (Most Frequent)

- Sometimes, we do not care about average or middle value. We want to know the most frequent data. This is especially useful for categorical data where mean and median do not apply.
- Mode is the value that appears most frequently in the dataset. A dataset can have no mode, single mode (unimodal) or multiple modes (bimodal, multimodal).
- **Example: Categorical Data**
    
    Bulb quality ratings: Good, Good, Good, Excellent, Fair, Good, Excellent
    
    Mode = Good (appears 4 times)

# Kurtosis

- While skewness tells about asymmetry of data, kurtosis tells about the tails. Specifically, how likely we have extreme values (outliers) compared to normal distribution?
- Kurtosis is a measure of "tailedness" of data distribution - whether data has heavy tail (more outliers) or light tail (less outliers) compared to normal distribution.

$$
\text{Kurtosis} = \frac {1}{n} \sum_{i=1}^{n} (\frac {x_i - \overline{x}}{\sigma})^4
$$

- The fourth power amplifies extreme deviations => Kurtosis is highly sensitive to outliers.

# Measures of Spread (Dispersion)

- We have looked at the **center** (mean, median) and **shape** (skewness, kurtosis). Now, we need to look at another critical piece of the puzzle - **spread**.
- Lets say, we have Company A and Company B with same average salary of \$100k. In Company A, everyone receives the same salary of \$100k. In Company B, CEO receives \$1M and everyone else earns \$10k. Even though average is the same, **spread tells the real story**.
- We can measure spread by calculating, on average, how far data points are from the center.

![rainbow](https://github.com/ancilcleetus/My-Learning-Journey/assets/25684256/839c3524-2a1d-4779-85a0-83c562e1e5e5)

# Variance and Standard Deviation

- Variance and Standard Deviation are the most commonly used measures of spread. They quantify how much each data point deviates from the mean.
- Variance is the average of the squared deviations from mean. Larger variance means larger spread.

$$
\text{Variance}, \sigma^2 = \frac {1}{n} \sum_{i=1}^{n} (x_i - \overline{x})^2
$$

- Standard Deviation is the square root of variance. It has the same units as the original data.

$$
\text{Standard Deviation}, \sigma = \sqrt{\frac {1}{n} \sum_{i=1}^{n} (x_i - \overline{x})^2}
$$

- **Example: Bulb Life Spans**

    Lifespans: 980, 1000, 1010, 1020, 1040 hours

    **Step 1: Calculate the mean**

    $\text{Mean} = (980 + 1000 + 1010 + 1020 + 1040)/5 = 5050/5 = 1010$

    **Step 2: Find each deviation from the mean**

    $980 - 1010 = -30$

    $1000 - 1010 = -10$

    $1010 - 1010 = 0$

    $1020 - 1010 = 10$

    $1040 - 1010 = 30$

    **Step 3: Find squares of deviation**

    $(-30)^2 = 900$

    $(-10)^2 = 100$

    $(0)^2 = 0$

    $(10)^2 = 100$

    $(30)^2 = 900$

    **Step 4: Calculate variance (average of squared deviations)**

    $\text{Variance} = (900 + 100 + 0 + 100 + 900)/5 = 400 \hspace{0.1cm} \text{hours}^2$

    **Step 5: Calculate standard deviation (square root of variance)**

    $\text{Standard Deviation} = \sqrt{400} = 20 \hspace{0.1cm} \text{hours}$

**Why Square the Deviations**

- If we do not square the deviations, positive and negative deviations may cancel out.
- Squaring penalizes large deviations heavily, making the metric more sensitive to outliers.

# Range and Interquartile Range (IQR)

- $\text{Range} = \text{Max} - \text{Min}$
    - Simplest measure of spread => difference between max & min values
    - Weakness: Extremely sensitive to outliers. Even one extreme value can inflate the range.
- $\text{Interquartile Range (IQR)} = Q_3 - Q_1$
    - Measures range of the middle 50% of data between $Q_1$ ($25^{th}$ percentile) and $Q_3$ ($75^{th}$ percentile)
    - Strength: Robust to outliers because it do not consider the extreme 25% on both ends.

**Outlier Detection**

A common rule to flag any data point as an outlier is if it falls below $Q_1 - 1.5 \times \text{IQR}$ or above $Q_3 + 1.5 \times \text{IQR}$. This is how box plots draw their "whiskers".

# Formulas Reference

| Metric | Formula | Use When |
| :----- | :------ | :------- |
| Mean | $\overline{x} = \frac {1}{n} \sum_{i=1}^{n} x_i$ | Symmetric data, no outliers |
| Median | Middle value (sorted) | Skewed data or when outliers present |
| Mode | Most frequent value | Categorical data or to identify peaks |
| Range | $\text{Max} - \text{Min}$ | Quick overview, no outliers |
| IQR | $Q_3 - Q_1$ | Robust spread against outliers, outlier detection |
| Variance | $\sigma^2 = \frac {1}{n} \sum_{i=1}^{n} (x_i - \overline{x})^2$ | To measure spread in Statistics & Machine Learning |
| Standard Deviation | $\sigma = \sqrt{\sigma^2}$ | Measure of spread in the same units as that of data |
