# Intro to Descriptive Statistics

What is Descriptive Statistics?
- Descriptive Statistics => tools to summarize and describe data
- Instead of looking at thousands of numbers, we extract key metrics that tell the story
- Eg: "This app has a 4.5 rating based on 10,000 reviews" => Opinions of thousands of people summarized into a single useful metric.

Essential Building Blocks
- Where is the "center" of your data? => **mean**, **median**, **mode**
- How spread out is your data? => **variance**, **standard deviation**
- Symmetry of your data => **Skewness**
- Tails in your data => **Kurtosis**

Why this matters for ML Engineers?
- **Make sense of data**: Quickly summarize thousands of numbers into a few key insights.
- **Spot problems**: Identify unusual values which can be errors or exceptions.
- **Compare fairly**: Understand if two datasets are truly different or just appear so.
- **Build intuition**: These concepts are the foundations for everything in data science.

# The Mean (Average)

- The most common way to summarize data is to find its "center" - a single value representing a "typical" data point. Mean is the most used measure of center.
- **Arithmetic Mean** (aka Average) is the sum of all values divided by the number of values:

$$
\overline{x} = \frac {1}{n}\sum_{i=1}^{n} x_i = \frac {x_1 + x_2 + ... + x_n}{n}
$$

- **Example: Bulb Lifespans**

    You test 5 bulbs with lifespans: 980, 1000, 1010, 1020, 1040 hours.

    Mean = (980 + 1000 + 1010 + 1020 + 1040) / 5 = 1010 hours

# The Median (Middle Value)

- The mean works well for symmetric data, but it has a weakness - outliers can pull it away from the true center. Median solves this weakness by **focusing on the position instead of the value**.
- Median is the middle value when data points are sorted. If we have even number of data points, median is the average of the two middle values:

$$
\text{Median} =
\begin{cases}
  x_{\frac{n+1}{2}}, & \text{if } n \text{ is odd} \\
  \frac{x_{\frac{n}{2}} + x_{\frac{n}{2} + 1}}{2}, & \text{if } n \text{ is even}
\end{cases}
$$

- **Example: With an Outlier**
    
    Lifespans: 500, 980, 1000, 1010, 1020 hours. (One bulb failed early.)

    Sorted: 500, 980, 1000, 1010, 1020
    Median: 1000 hours (middle value)
    Mean: 902 hours (pulled down by the outlier)
